```python
import os

os.chdir(r"C:\Users\2k23c\HandyHub")

file_path = "templates/index.html"

with open(file_path, "r", encoding="utf-8") as f:
    html = f.read()

html = html.replace(
    '<button class="close" onclick="closeModal()">×</button>',
    '<button type="button" class="close" onclick="document.getElementById(\'modal\').classList.add(\'hidden\')">×</button>'
)

with open(file_path, "w", encoding="utf-8") as f:
    f.write(html)

print("Close button fixed successfully.")
```

```python
with open("static/js/script.js", "a", encoding="utf-8") as f:
    f.write("""
    
document.getElementById("modal")?.addEventListener("click", function(event) {
    if (event.target === this) {
        this.classList.add("hidden");
    }
});

""")

print("Outside-click close functionality added.")
```
