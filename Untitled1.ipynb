```python
import os

# Make sure we are in HandyHub
os.chdir(r"C:\Users\2k23c\HandyHub")

# Create folders
os.makedirs("templates", exist_ok=True)
os.makedirs("static/css", exist_ok=True)
os.makedirs("static/js", exist_ok=True)

# -----------------------------
# 1. index.html
# -----------------------------
index_html = """
<!DOCTYPE html>
<html>
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>HandyHub - Local Services</title>
    <link rel="stylesheet" href="/static/css/style.css">
</head>

<body>

<nav>
    <a class="brand" href="/">HandyHub</a>

    <div class="navlinks">
        <a href="/">Home</a>

        {% if user %}
            <a href="/dashboard">Dashboard</a>
            <button class="linkbtn" onclick="logout()">Logout</button>
        {% else %}
            <a href="/login">Login</a>
            <a class="btn small" href="/register">Register</a>
        {% endif %}
    </div>
</nav>

<section class="hero">
    <div>
        <span class="pill">LOCAL SERVICES • MADE SIMPLE</span>

        <h1>Find trusted local professionals.</h1>

        <p>
            Discover electricians, plumbers, carpenters,
            cleaners and other professionals in one place.
        </p>

        <a class="btn" href="#providers">Find a Service</a>
    </div>
</section>

<section class="section" id="providers">

    <div class="sectionhead">

        <div>
            <h2>Available Professionals</h2>
            <p>Search by service, provider or city.</p>
        </div>

        <div class="filters">

            <input
                id="search"
                placeholder="Search service or provider"
            >

            <input
                id="city"
                placeholder="City"
            >

            <select id="service">
                <option value="">All Services</option>
            </select>

            <button class="btn" onclick="loadProviders()">
                Search
            </button>

        </div>

    </div>

    <div id="providerGrid" class="grid">
        Loading providers...
    </div>

</section>

<div id="modal" class="modal hidden">

    <div class="modalbox">

        <button class="close" onclick="closeModal()">×</button>

        <h2>Request Service</h2>

        <div id="modalInfo"></div>

        <textarea
            id="description"
            placeholder="Describe what you need..."
        ></textarea>

        <button class="btn" onclick="submitRequest()">
            Send Request
        </button>

    </div>

</div>

<footer>
    © 2026 HandyHub | Local Services Platform
</footer>

<script src="/static/js/script.js"></script>

</body>
</html>
"""

# -----------------------------
# 2. login.html
# -----------------------------
login_html = """
<!DOCTYPE html>
<html>
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Login - HandyHub</title>
    <link rel="stylesheet" href="/static/css/style.css">
</head>

<body>

<nav>
    <a class="brand" href="/">HandyHub</a>

    <div class="navlinks">
        <a href="/">Home</a>
        <a href="/register">Register</a>
    </div>
</nav>

<main class="auth">

    <div class="authbox">

        <h1>Welcome Back</h1>

        <p>Login to your HandyHub account.</p>

        <form id="loginForm">

            <input
                name="email"
                type="email"
                placeholder="Email"
                required
            >

            <input
                name="password"
                type="password"
                placeholder="Password"
                required
            >

            <button class="btn full">
                Login
            </button>

            <div id="message" class="message"></div>

        </form>

    </div>

</main>

<script src="/static/js/script.js"></script>

</body>
</html>
"""

# -----------------------------
# 3. register.html
# -----------------------------
register_html = """
<!DOCTYPE html>
<html>
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Register - HandyHub</title>
    <link rel="stylesheet" href="/static/css/style.css">
</head>

<body>

<nav>
    <a class="brand" href="/">HandyHub</a>

    <div class="navlinks">
        <a href="/">Home</a>
        <a href="/login">Login</a>
    </div>
</nav>

<main class="auth">

    <div class="authbox wide">

        <h1>Create Account</h1>

        <p>Join HandyHub as a customer or service provider.</p>

        <form id="registerForm">

            <div class="twocol">

                <input
                    name="name"
                    placeholder="Full Name"
                    required
                >

                <input
                    name="email"
                    type="email"
                    placeholder="Email"
                    required
                >

            </div>

            <div class="twocol">

                <input
                    name="password"
                    type="password"
                    minlength="6"
                    placeholder="Password"
                    required
                >

                <input
                    name="phone"
                    placeholder="Phone"
                >

            </div>

            <div class="twocol">

                <input
                    name="city"
                    placeholder="City"
                >

                <select name="role" id="role">

                    <option value="customer">
                        Customer
                    </option>

                    <option value="provider">
                        Service Provider
                    </option>

                </select>

            </div>

            <div id="providerFields" class="hidden">

                <div class="twocol">

                    <select
                        name="service_id"
                        id="registerService"
                    >
                        <option value="">
                            Select Service
                        </option>
                    </select>

                    <input
                        name="experience"
                        type="number"
                        min="0"
                        placeholder="Experience (years)"
                    >

                </div>

                <input
                    name="price"
                    type="number"
                    min="0"
                    placeholder="Starting Price"
                >

            </div>

            <button class="btn full">
                Create Account
            </button>

            <div id="message" class="message"></div>

        </form>

    </div>

</main>

<script src="/static/js/script.js"></script>

</body>
</html>
"""

# -----------------------------
# 4. dashboard.html
# -----------------------------
dashboard_html = """
<!DOCTYPE html>
<html>
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Dashboard - HandyHub</title>
    <link rel="stylesheet" href="/static/css/style.css">
</head>

<body>

<nav>

    <a class="brand" href="/">
        HandyHub
    </a>

    <div class="navlinks">

        <a href="/">
            Services
        </a>

        <button class="linkbtn" onclick="logout()">
            Logout
        </button>

    </div>

</nav>

<main class="section">

    <div class="welcome">

        <span class="pill">
            DASHBOARD
        </span>

        <h1>
            Hello, {{ user.name }} 👋
        </h1>

        <p>
            {{ user.role|capitalize }}
            •
            {{ user.city or "City not set" }}
        </p>

    </div>

    <h2>Your Service Requests</h2>

    <div id="requests" class="requests">
        Loading...
    </div>

</main>

<footer>
    HandyHub
</footer>

<script src="/static/js/script.js"></script>

</body>
</html>
"""

# -----------------------------
# 5. CSS
# -----------------------------
css = """
* {
    box-sizing: border-box;
}

body {
    margin: 0;
    font-family: Arial, sans-serif;
    background: #f7f8fc;
    color: #172033;
}

a {
    text-decoration: none;
    color: inherit;
}

nav {
    height: 72px;
    background: white;
    border-bottom: 1px solid #e5e7eb;
    display: flex;
    align-items: center;
    justify-content: space-between;
    padding: 0 6%;
}

.brand {
    font-size: 25px;
    font-weight: 800;
}

.navlinks {
    display: flex;
    gap: 20px;
    align-items: center;
}

.linkbtn {
    border: 0;
    background: none;
    cursor: pointer;
    font-size: 15px;
}

.btn {
    display: inline-block;
    border: 0;
    border-radius: 10px;
    padding: 12px 20px;
    background: #172033;
    color: white;
    font-weight: bold;
    cursor: pointer;
}

.btn.small {
    padding: 9px 15px;
}

.btn.full {
    width: 100%;
}

.hero {
    min-height: 470px;
    display: flex;
    align-items: center;
    padding: 70px 8%;
    background: linear-gradient(
        120deg,
        #e9f1ff,
        #fff5e8
    );
}

.hero h1 {
    font-size: 55px;
    max-width: 700px;
    margin: 18px 0;
}

.hero p {
    font-size: 19px;
    max-width: 600px;
    color: #687185;
    line-height: 1.6;
}

.pill {
    font-size: 12px;
    font-weight: bold;
    background: white;
    padding: 8px 12px;
    border-radius: 20px;
    border: 1px solid #ddd;
}

.section {
    padding: 55px 7%;
    max-width: 1400px;
    margin: auto;
}

.sectionhead {
    display: flex;
    justify-content: space-between;
    gap: 20px;
    align-items: end;
    margin-bottom: 25px;
}

.section h2 {
    font-size: 30px;
}

.filters {
    display: flex;
    gap: 8px;
    flex-wrap: wrap;
}

.filters input,
.filters select,
.authbox input,
.authbox select,
textarea {
    border: 1px solid #dce1ea;
    border-radius: 9px;
    padding: 12px;
    background: white;
    font-size: 14px;
}

.grid {
    display: grid;
    grid-template-columns:
        repeat(auto-fit, minmax(270px, 1fr));
    gap: 18px;
}

.card {
    background: white;
    border: 1px solid #e5e8ef;
    border-radius: 15px;
    padding: 22px;
    box-shadow: 0 5px 20px rgba(0,0,0,.03);
}

.avatar {
    width: 52px;
    height: 52px;
    border-radius: 50%;
    display: grid;
    place-items: center;
    background: #eef2ff;
    font-size: 20px;
    font-weight: bold;
}

.card h3 {
    margin-bottom: 5px;
}

.meta {
    color: #687185;
    font-size: 14px;
    margin: 7px 0;
}

.price {
    font-weight: bold;
    margin: 15px 0;
}

.card .btn {
    width: 100%;
}

.auth {
    min-height: calc(100vh - 72px);
    display: grid;
    place-items: center;
    padding: 30px;
}

.authbox {
    background: white;
    width: min(440px, 100%);
    padding: 35px;
    border: 1px solid #e4e7ee;
    border-radius: 18px;
}

.authbox.wide {
    width: min(650px, 100%);
}

.authbox form {
    display: grid;
    gap: 13px;
}

.twocol {
    display: grid;
    grid-template-columns: 1fr 1fr;
    gap: 12px;
}

.hidden {
    display: none;
}

.message {
    padding: 8px;
    font-size: 14px;
}

.modal {
    position: fixed;
    inset: 0;
    background: rgba(0,0,0,.5);
    display: grid;
    place-items: center;
    z-index: 10;
    padding: 20px;
}

.modalbox {
    background: white;
    width: min(500px,100%);
    padding: 28px;
    border-radius: 16px;
    position: relative;
}

.close {
    position: absolute;
    right: 15px;
    top: 10px;
    border: 0;
    background: none;
    font-size: 28px;
    cursor: pointer;
}

.modalbox textarea {
    width: 100%;
    height: 130px;
    margin: 15px 0;
    resize: vertical;
}

.modalbox .btn {
    width: 100%;
}

.request {
    background: white;
    border: 1px solid #e3e6ed;
    border-radius: 13px;
    padding: 18px;
    margin: 12px 0;
}

.status {
    display: inline-block;
    padding: 5px 9px;
    border-radius: 20px;
    background: #eef2ff;
    font-size: 12px;
    font-weight: bold;
}

footer {
    text-align: center;
    padding: 35px;
    color: #7b8495;
    border-top: 1px solid #e5e8ee;
}

@media(max-width:750px) {

    .hero h1 {
        font-size: 40px;
    }

    .sectionhead {
        display: block;
    }

    .filters {
        margin-top: 20px;
    }

    .filters > * {
        width: 100%;
    }

    .twocol {
        grid-template-columns: 1fr;
    }
}
"""

# -----------------------------
# 6. JavaScript
# -----------------------------
js = """
let selectedProvider = null;

async function api(url, options = {}) {

    const response = await fetch(url, {
        headers: {
            "Content-Type": "application/json"
        },
        ...options
    });

    const data = await response.json().catch(() => ({}));

    if (!response.ok) {
        throw new Error(
            data.error || "Something went wrong"
        );
    }

    return data;
}


async function loadServices() {

    const data = await api("/api/services");

    const serviceSelect =
        document.getElementById("service");

    if (serviceSelect) {

        data.forEach(service => {

            serviceSelect.insertAdjacentHTML(
                "beforeend",
                `<option value="${service.id}">
                    ${service.name}
                </option>`
            );

        });
    }

    const registerService =
        document.getElementById("registerService");

    if (registerService) {

        registerService.innerHTML =
            '<option value="">Select Service</option>' +
            data.map(service =>
                `<option value="${service.id}">
                    ${service.name}
                </option>`
            ).join("");

    }
}


async function loadProviders() {

    const grid =
        document.getElementById("providerGrid");

    if (!grid) {
        return;
    }

    grid.innerHTML = "Loading providers...";

    const q =
        encodeURIComponent(
            document.getElementById("search")?.value || ""
        );

    const city =
        encodeURIComponent(
            document.getElementById("city")?.value || ""
        );

    const service =
        document.getElementById("service")?.value || "";

    try {

        const data = await api(
            `/api/providers?q=${q}&city=${city}&service=${service}`
        );

        if (!data.length) {

            grid.innerHTML =
                "<p>No matching providers found.</p>";

            return;
        }

        grid.innerHTML = data.map(provider => `

            <article class="card">

                <div class="avatar">
                    ${provider.name.charAt(0).toUpperCase()}
                </div>

                <h3>
                    ${escapeHtml(provider.name)}
                </h3>

                <div class="meta">
                    ${escapeHtml(provider.service)}
                    •
                    ${escapeHtml(provider.city || "")}
                </div>

                <div class="meta">
                    ${provider.experience}
                    years experience
                </div>

                <div class="price">
                    ₹${Number(provider.price)
                        .toLocaleString("en-IN")}
                    starting
                </div>

                <button
                    class="btn"
                    onclick='openRequest(${JSON.stringify(provider)})'
                >
                    Request Service
                </button>

            </article>

        `).join("");

    } catch (error) {

        grid.innerHTML =
            `<p>${escapeHtml(error.message)}</p>`;

    }
}


function openRequest(provider) {

    selectedProvider = provider;

    document.getElementById(
        "modalInfo"
    ).innerHTML = `
        <p>
            <strong>
                ${escapeHtml(provider.name)}
            </strong>
            —
            ${escapeHtml(provider.service)}
            in
            ${escapeHtml(provider.city || "")}
        </p>
    `;

    document.getElementById(
        "description"
    ).value = "";

    document.getElementById(
        "modal"
    ).classList.remove("hidden");
}


function closeModal() {

    document.getElementById(
        "modal"
    ).classList.add("hidden");

}


async function submitRequest() {

    if (!selectedProvider) {
        return;
    }

    const description =
        document.getElementById(
            "description"
        ).value.trim();

    if (!description) {

        alert(
            "Please describe the service you need."
        );

        return;
    }

    try {

        const result = await api(
            "/api/requests",
            {
                method: "POST",

                body: JSON.stringify({
                    provider_id:
                        selectedProvider.id,

                    service_id:
                        selectedProvider.service_id,

                    description:
                        description
                })
            }
        );

        alert(
            result.message +
            " (#" +
            result.request_id +
            ")"
        );

        closeModal();

    } catch (error) {

        if (
            error.message.includes(
                "Authentication"
            )
        ) {

            window.location.href =
                "/login";

        } else {

            alert(error.message);

        }
    }
}


async function logout() {

    await api(
        "/api/logout",
        {
            method: "POST"
        }
    );

    window.location.href = "/";
}


function escapeHtml(value) {

    return String(value ?? "").replace(
        /[&<>"']/g,

        character => ({

            "&": "&amp;",
            "<": "&lt;",
            ">": "&gt;",
            '"': "&quot;",
            "'": "&#039;"

        }[character])
    );
}


async function loadRequests() {

    const box =
        document.getElementById("requests");

    if (!box) {
        return;
    }

    try {

        const data =
            await api("/api/requests");

        if (!data.length) {

            box.innerHTML =
                "<p>No service requests yet.</p>";

            return;
        }

        box.innerHTML = data.map(request => {

            const person =
                request.provider_name ||
                request.customer_name ||
                "User";

            return `

                <div class="request">

                    <strong>
                        ${escapeHtml(request.service)}
                    </strong>

                    <span class="status">
                        ${escapeHtml(request.status)}
                    </span>

                    <p>
                        ${escapeHtml(
                            request.description
                        )}
                    </p>

                    <small>
                        With:
                        ${escapeHtml(person)}
                    </small>

                    <br><br>

                    ${
                        request.status === "Pending"
                        ?
                        `<button
                            class="btn small"
                            onclick="changeStatus(
                                ${request.id},
                                'Cancelled'
                            )"
                        >
                            Cancel
                        </button>`
                        :
                        ""
                    }

                    ${
                        request.customer_name
                        ?
                        `<button
                            class="btn small"
                            onclick="changeStatus(
                                ${request.id},
                                'Accepted'
                            )"
                        >
                            Accept
                        </button>

                        <button
                            class="btn small"
                            onclick="changeStatus(
                                ${request.id},
                                'Rejected'
                            )"
                        >
                            Reject
                        </button>`
                        :
                        ""
                    }

                    ${
                        request.status === "Accepted"
                        ?
                        `<button
                            class="btn small"
                            onclick="changeStatus(
                                ${request.id},
                                'Completed'
                            )"
                        >
                            Complete
                        </button>`
                        :
                        ""
                    }

                </div>
            `;

        }).join("");

    } catch (error) {

        box.innerHTML =
            `<p>${escapeHtml(error.message)}</p>`;

    }
}


async function changeStatus(id, status) {

    try {

        await api(
            `/api/requests/${id}`,
            {
                method: "PATCH",

                body: JSON.stringify({
                    status: status
                })
            }
        );

        loadRequests();

    } catch (error) {

        alert(error.message);

    }
}


document.addEventListener(
    "DOMContentLoaded",
    () => {

        loadServices().catch(console.error);

        loadProviders().catch(console.error);

        loadRequests();

        const role =
            document.getElementById("role");

        if (role) {

            role.addEventListener(
                "change",
                () => {

                    document
                        .getElementById(
                            "providerFields"
                        )
                        .classList.toggle(
                            "hidden",
                            role.value !== "provider"
                        );

                }
            );

        }


        const loginForm =
            document.getElementById(
                "loginForm"
            );

        if (loginForm) {

            loginForm.addEventListener(
                "submit",
                async event => {

                    event.preventDefault();

                    const formData =
                        new FormData(loginForm);

                    try {

                        await api(
                            "/api/login",
                            {
                                method: "POST",

                                body: JSON.stringify(
                                    Object.fromEntries(
                                        formData
                                    )
                                )
                            }
                        );

                        window.location.href =
                            "/dashboard";

                    } catch (error) {

                        document
                            .getElementById(
                                "message"
                            )
                            .textContent =
                            error.message;

                    }

                }
            );

        }


        const registerForm =
            document.getElementById(
                "registerForm"
            );

        if (registerForm) {

            registerForm.addEventListener(
                "submit",
                async event => {

                    event.preventDefault();

                    const formData =
                        new FormData(
                            registerForm
                        );

                    try {

                        await api(
                            "/api/register",
                            {
                                method: "POST",

                                body: JSON.stringify(
                                    Object.fromEntries(
                                        formData
                                    )
                                )
                            }
                        );

                        window.location.href =
                            "/dashboard";

                    } catch (error) {

                        document
                            .getElementById(
                                "message"
                            )
                            .textContent =
                            error.message;

                    }

                }
            );

        }

    }
);
"""

# Write all files
with open("templates/index.html", "w", encoding="utf-8") as f:
    f.write(index_html)

with open("templates/login.html", "w", encoding="utf-8") as f:
    f.write(login_html)

with open("templates/register.html", "w", encoding="utf-8") as f:
    f.write(register_html)

with open("templates/dashboard.html", "w", encoding="utf-8") as f:
    f.write(dashboard_html)

with open("static/css/style.css", "w", encoding="utf-8") as f:
    f.write(css)

with open("static/js/script.js", "w", encoding="utf-8") as f:
    f.write(js)

print("Frontend files created successfully!")
print()
print("templates:")
print(os.listdir("templates"))
print()
print("static/css:")
print(os.listdir("static/css"))
print()
print("static/js:")
print(os.listdir("static/js"))
```
